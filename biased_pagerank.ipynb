```python
import ppipy
import matplotlib.pyplot as plt
%env NX_CUGRAPH_AUTOCONFIG=True
import networkx as nx
```

```python
G = ppipy.initialise_yeast_PPI_network()
```

```python
nx.community.leiden_communities(G)
```

```python
ranks, communities = ppipy.bias_pagerank(G, resolution=50, important_nodes=dict([['YMR205C', 20.0], ['YFL025C', 1.0], 
                        ['YPL031C', 1.0], ['YPR074C', 1.0]]), alpha= 0.85, beta=1.0)
# Green YMR205C, Red YER003C, YDR516C
```

```python
for j in range(len(ranks)):
    scores = []
    for i in ranks[j]:
        scores.append(i[1])
    plt.plot(scores)
```

```python
fig, ax = plt.subplots()
fig.set_dpi(250)
important_nodes = dict([['YMR205C', 100.0], ['YFL025C', 50.0], ['YPL031C', 90.0], ['YPR074C', 70.0]])
i = 0
community = communities[i]
most_important = [ranks[i][1][0], ranks[i][2][0], ranks[i][3][0]]
colour_map = []
for node in community:
    if node in most_important:
        colour_map.append('red')
    elif node in important_nodes:
        colour_map.append('green')
    elif node == 'YML012W':
        colour_map.append('yellow')
        print('found')
    else:
        colour_map.append('blue')
nx.draw(community, pos=nx.spring_layout(community, k=2), with_labels=True, node_color=colour_map, width=0.2, ax=ax)
```

# Statistics

```python
# no bias
n = 50
m = 3
rank_stats = []
for i in range(0, n):
    ranks, communities = ppipy.bias_pagerank(G, resolution=50, important_nodes=dict([['YMR205C', 1.0], ['YFL025C', 1.0], 
                        ['YPL031C', 1.0], ['YPR074C', 1.0]]), alpha= 0.85, beta=1.0)
    for j in range(len(ranks)):
        if ranks[j][0][0] == 'YMR205C':
            for k in range(1, m+1):
                rank_stats.append(ranks[j][k][0])
#                 print(ranks[j][k][0])
```

```python
fig, ax = plt.subplots(figsize=(10,5))
ax.hist(rank_stats, density=True)
fig.set_dpi(300)
```
